```python
##### Importing libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

##### Dir path
datos_path = os.path.join(
    os.getcwd(),
    'Datos',
    'Limpios'
)

##### File names
nom_1 = 'visitantes_internacionales.csv'
nom_2 = 'visitantes_sitios_turisticos.csv'
nom_3 = 'inventario_recursos_turisticos.csv'
```

##### EDA Datos 1: Visitantes internacionales

```python
##### Loading data
df_1 = pd.read_csv(f'{datos_path}/{nom_1}')
df_1.head()
```

```python
##### Quick summary
df_1.info()

##### Eliminating unnecessary columns
df_1 = df_1.drop(columns=['FECHA_CORTE', 'MES', 'TIPO_VISITANTE'])
```

```python
###### Separating variables: categorical vs numeric
df_1.info()
cuant_vars_1 = df_1._get_numeric_data().columns.values
    # Cuantitativos -> Numerico (discretos o continuo)
cual_vars_1 = list(set(df_1.columns)- set(cuant_vars_1))
    # Cualitativa -> Nominal o ordinal
```

```python
##### Analyzing data
#### Cuant
display(df_1[cuant_vars_1].describe())

#### Cual
for col in cual_vars_1:
    cat_num = df_1[col].value_counts()[:20]
    print("graph for %s: total = %d" % (col, len(cat_num)))
    chart = sns.barplot(x=cat_num.index, y=cat_num)
    chart.set_xticklabels(chart.get_xticklabels(), rotation=90)
    plt.show()
```

```python
##### Correlacion
cmap = sns.diverging_palette(220, 10, as_cmap=True)
sns.heatmap(df_1[cuant_vars_1].corr(),vmax=.3, center=0, cmap=cmap,
            square=True, linewidths=.5, cbar_kws={"shrink": .5})
```

        * International Airport (Lima)
            - Map: number 106

* OCM
    * Names of the OCMs

    - The following variations exist:
        * Puno: 66,528 = 33,264 * 2
        * Lima and Callao: 33,264

* NUMBER_VISITORS
    - This is an odd column
    - (min, mean, median, max) = (0, 100, 0, 94883)
    - This data is extremely right-skewed
        * Evidence: mean = 100 and median = 0
        * This means that most of the data will be outliers

        * 2,261 records have ‘NUMBER_VISITORS’ in the range (50, 100]

        * 1 visitor: 7,896 times
        * 2 visitors: 4,056 times
        * 3 visitors: 2,535 times
        * 4–7 visitors: between 1,000 and 2,000 times
        * 8–12 visitors: between 500 and 1,000 times
        * 13+ visitors: fewer than 500 times
    - In other words, most tourist sites have only a few visitors
        * Also, this data spans 7 years
        * But that doesn’t change the fact that most have 0 visitors
